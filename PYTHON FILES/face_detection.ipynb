```python
s=cv2.imread('d.jpg')
f=cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
def detect_face(img):
    img1=img.copy()
    face=f.detectMultiScale(img1)
    for (x,y,w,h) in face:
        cv2.rectangle(img1,(x,y),(x+w,y+h),(255,255,0),3)
    return img1
```

```python
r=detect_face(s)
plt.imshow(r)
```
