```python
# get the uploaded filename with Library
import PyPDF2
pdf_path = next(iter(uploaded))
```

```python
# Extract the Texts from pdf
reader = PyPDF2.PdfReader(pdf_path)

pdf_text = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        pdf_text += text + " "

print("PDF Extracted Successfully!")
print(pdf_text[:100000])  # show first 1000 chars
```

Text PreProcessing

```python
#STARTING TEXT (resume extracted content)
resume_raw_text = pdf_text
```

```python
#Convert to lowercase
resume_lower = resume_raw_text.lower()
```

```python
#Remove punctuation
resume_no_punct = re.sub(r'[^\w\s]', '', resume_lower)
```

```python
#Remove numbers
resume_no_numbers = re.sub(r'\d+', '', resume_no_punct)
```

```python
#Remove extra spaces
resume_clean_text = re.sub(r'\s+', ' ', resume_no_numbers).strip()
```

```python
print("CLEANED RESUME TEXT (first 30000 chars):\n")
print(resume_clean_text[:30000])
```

Tokenizing the Texts Extracted

```python
resume_tokens = word_tokenize(resume_clean_text)
```

```python
print("\nRESUME TOKENS (first 40):\n")
print(resume_tokens[:40])
```

Text Summarizations using transformers

```python
!pip install transformers
from transformers import pipeline
```

```python
# load summarization model
summarizer = pipeline("summarization",model="facebook/bart-large-cnn")
```

```python
# BART has a limit of ~1024 tokens → break long text if needed
resume_text_for_summary = resume_clean_text[:2000]   # safe trimming
```

```python
summary = summarizer(
    resume_text_for_summary,
    max_length=180,     # length of summary
    min_length=60,      # minimum summary
    do_sample=False
)[0]['summary_text']

print("RESUME SUMMARY:\n")
print(summary)
```
